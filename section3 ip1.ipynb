### EUCLIDEAN DISTANCE

```python
import cv2
import math

img = cv2.imread(r"D:\image_processing_lab\lenna.png", 0)

P = (50, 50)
Q = (80, 90)

distance = math.sqrt(
    (P[0] - Q[0])**2 +
    (P[1] - Q[1])**2
)

print("Euclidean Distance =", distance)
```

### CITY-BLOCK DISTANCE

```python
import cv2
img = cv2.imread(r"D:\image_processing_lab\lenna.png", 0)

P = (50, 50)
Q = (80, 90)

distance = abs(P[0] - Q[0]) + abs(P[1] - Q[1])

print("City-block Distance =", distance)
```

### CHESSBOARD DISTANCE

```python
import cv2
img = cv2.imread("binary_image.png", 0)

P = (50, 50)
Q = (80, 90)

distance = max(
    abs(P[0] - Q[0]),
    abs(P[1] - Q[1])
)

print("Chessboard Distance =", distance)
```
