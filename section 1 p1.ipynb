```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
img = cv2.imread(
    r"D:\image_processing_lab\lenna.png",
    cv2.IMREAD_GRAYSCALE
)

# -------------------------------
# a. Increase brightness
# -------------------------------
brightness_increased = cv2.add(img, 50)

# -------------------------------
# b. Decrease brightness
# -------------------------------
brightness_decreased = cv2.subtract(img, 50)

# -------------------------------
# c. Increase contrast
# -------------------------------
# Formula: new_pixel = alpha * pixel
alpha_increase = 1.5
contrast_increased = cv2.convertScaleAbs(
    img, alpha=alpha_increase, beta=0
)

# -------------------------------
# d. Decrease contrast
# -------------------------------
alpha_decrease = 0.5
contrast_decreased = cv2.convertScaleAbs(
    img, alpha=alpha_decrease, beta=0
)

# -------------------------------
# e. Generate negative image
# -------------------------------
negative = 255 - img

# -------------------------------
# f. Binary thresholding
# -------------------------------
threshold = 127
_, binary = cv2.threshold(
    img, threshold, 255, cv2.THRESH_BINARY
)

# -------------------------------
# Display all images
# -------------------------------
images = [
    img,
    brightness_increased,
    brightness_decreased,
    contrast_increased,
    contrast_decreased,
    negative,
    binary
]

titles = [
    "Original",
    "Increased Brightness",
    "Decreased Brightness",
    "Increased Contrast",
    "Decreased Contrast",
    "Negative",
    "Binary Threshold"
]

plt.figure(figsize=(32, 32))

for i in range(len(images)):
    plt.subplot(2, 4, i + 1)
    plt.imshow(images[i], cmap="gray")
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```
