```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread(
    r"D:\image_processing_lab\lenna.png"
)

# Get image dimensions
rows, cols = img.shape[:2]

# --------------------------------
# a. Translation
# --------------------------------
# Move image 100 pixels right and 50 pixels down
tx = 100
ty = 50

translation_matrix = np.float32([
    [1, 0, tx],
    [0, 1, ty]
])

translated = cv2.warpAffine(
    img,
    translation_matrix,
    (cols, rows)
)

# --------------------------------
# b. Scaling
# --------------------------------
# Increase image size by 1.5 times
scaled = cv2.resize(
    img,
    None,
    fx=1.5,
    fy=1.5,
    interpolation=cv2.INTER_LINEAR
)

# --------------------------------
# c. Rotation
# --------------------------------
# Rotate image by 45 degrees
center = (cols // 2, rows // 2)

rotation_matrix = cv2.getRotationMatrix2D(
    center,
    45,
    1
)

rotated = cv2.warpAffine(
    img,
    rotation_matrix,
    (cols, rows)
)

# --------------------------------
# d. Horizontal Reflection
# --------------------------------
# Flip around horizontal axis
horizontal_reflection = cv2.flip(img, 0)

# --------------------------------
# e. Vertical Reflection
# --------------------------------
# Flip around vertical axis
vertical_reflection = cv2.flip(img, 1)

# --------------------------------
# Display all images
# --------------------------------
images = [
    img,
    translated,
    scaled,
    rotated,
    horizontal_reflection,
    vertical_reflection
]

titles = [
    "Original",
    "Translation",
    "Scaling",
    "Rotation",
    "Horizontal Reflection",
    "Vertical Reflection"
]

plt.figure(figsize=(12, 8))

for i in range(len(images)):
    plt.subplot(2, 3, i + 1)

    # Convert BGR to RGB for Matplotlib
    if len(images[i].shape) == 3:
        display_img = cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB)
    else:
        display_img = images[i]

    plt.imshow(display_img)
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()

```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread(
    r"D:\image_processing_lab\lenna.png"
)

# Convert BGR to RGB for displaying with Matplotlib
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Get original dimensions
rows, cols = img.shape[:2]

# ==========================================
# ZOOMING - Increase image size by 2 times
# ==========================================

zoom_nearest = cv2.resize(
    img,
    None,
    fx=2,
    fy=2,
    interpolation=cv2.INTER_NEAREST
)

zoom_bilinear = cv2.resize(
    img,
    None,
    fx=2,
    fy=2,
    interpolation=cv2.INTER_LINEAR
)

zoom_bicubic = cv2.resize(
    img,
    None,
    fx=2,
    fy=2,
    interpolation=cv2.INTER_CUBIC
)

# ==========================================
# SHRINKING - Reduce image size by 0.5 times
# ==========================================

shrink_nearest = cv2.resize(
    img,
    None,
    fx=0.5,
    fy=0.5,
    interpolation=cv2.INTER_NEAREST
)

shrink_bilinear = cv2.resize(
    img,
    None,
    fx=0.5,
    fy=0.5,
    interpolation=cv2.INTER_LINEAR
)

shrink_bicubic = cv2.resize(
    img,
    None,
    fx=0.5,
    fy=0.5,
    interpolation=cv2.INTER_CUBIC
)

# ==========================================
# Display Zooming Results
# ==========================================

plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(cv2.cvtColor(zoom_nearest, cv2.COLOR_BGR2RGB))
plt.title("Zoom - Nearest Neighbor")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(cv2.cvtColor(zoom_bilinear, cv2.COLOR_BGR2RGB))
plt.title("Zoom - Bilinear")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(cv2.cvtColor(zoom_bicubic, cv2.COLOR_BGR2RGB))
plt.title("Zoom - Bicubic")
plt.axis("off")

plt.subplot(2, 3, 5) 
plt.imshow(cv2.cvtColor(shrink_nearest, cv2.COLOR_BGR2RGB)) 
plt.title("Shrink - Nearest Neighbor") 
plt.axis("off") 

plt.subplot(2, 3, 6) 
plt.imshow(cv2.cvtColor(shrink_bilinear, cv2.COLOR_BGR2RGB)) 
plt.title("Shrink - Bilinear") 
plt.axis("off") 
plt.tight_layout() 
plt.show()

plt.figure(figsize=(5, 5)) 
plt.imshow(cv2.cvtColor(shrink_bicubic, cv2.COLOR_BGR2RGB)) 
plt.title("Shrink - Bicubic") 
plt.axis("off") 
plt.show()
```
